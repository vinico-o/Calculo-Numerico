```python
import numpy as np
```

```python
def diferenciacao_numerica(x0, f, df, Df, delta):

    tamanho = len(delta)

    df_progressiva = np.zeros(tamanho)
    df_regressiva = np.zeros(tamanho)
    df_centrada = np.zeros(tamanho)
    segunda_derivada = np.zeros(tamanho)

    erro_df_progressiva = np.zeros(tamanho)
    erro_df_regressiva = np.zeros(tamanho)
    erro_df_centrada = np.zeros(tamanho)
    erro_segunda_derivada = np.zeros(tamanho)

    for i in range(tamanho):

        df_progressiva[i] = (f(x0 + delta[i]) - f(x0)) / delta[i]
        df_regressiva[i] = (f(x0) - f(x0 - delta[i])) / delta[i]
        df_centrada[i] = (f(x0 + delta[i]) - f(x0 - delta[i])) / (2 * delta[i])
        segunda_derivada[i] = (f(x0 + delta[i]) - 2 * f(x0) + f(x0 - delta[i])) / (delta[i] ** 2)
        erro_df_progressiva[i] = abs(df(x0) - df_progressiva[i]) / df(x0)
        erro_df_regressiva[i] = abs(df(x0) - df_regressiva[i]) / df(x0)
        erro_df_centrada[i] = abs(df(x0) - df_centrada[i]) / df(x0)
        erro_segunda_derivada[i] = abs(Df(x0) - segunda_derivada[i]) / Df(x0)

    return [df_progressiva, erro_df_progressiva, df_regressiva,erro_df_regressiva,
                df_centrada, erro_df_centrada, segunda_derivada, erro_segunda_derivada]
```

```python
def diferenciacao_mista(x0, y0, f, dfx, dfy, Dfx, Dfy, Dfxy, deltax, deltay):

    tamanho = len(deltax)

    df_progressiva_x = np.zeros(tamanho)
    df_regressiva_x = np.zeros(tamanho)
    df_centrada_x = np.zeros(tamanho)
    segunda_derivada_x = np.zeros(tamanho)

    df_progressiva_y = np.zeros(tamanho)
    df_regressiva_y = np.zeros(tamanho)
    df_centrada_y = np.zeros(tamanho)
    segunda_derivada_y = np.zeros(tamanho)

    segunda_derivada_xy = np.zeros(tamanho)

    erro_df_progressiva_x = np.zeros(tamanho)
    erro_df_regressiva_x = np.zeros(tamanho)
    erro_df_centrada_x = np.zeros(tamanho)
    erro_segunda_derivada_x = np.zeros(tamanho)

    erro_df_progressiva_y = np.zeros(tamanho)
    erro_df_regressiva_y = np.zeros(tamanho)
    erro_df_centrada_y = np.zeros(tamanho)
    erro_segunda_derivada_y = np.zeros(tamanho)

    erro_segunda_derivada_xy = np.zeros(tamanho)

    for i in range(tamanho):

        df_progressiva_x[i] = (f(x0 + deltax[i], y0) - f(x0, y0)) / deltax[i]
        df_regressiva_x[i] = (f(x0, y0) - f(x0 - deltax[i], y0)) / deltax[i]
        df_centrada_x[i] = (f(x0 + deltax[i], y0) - f(x0 - deltax[i], y0)) / (2 * deltax[i])
        segunda_derivada_x[i] = (f(x0 + deltax[i], y0) - 2 * f(x0, y0) + f(x0 - deltax[i], y0)) / (deltax[i] ** 2)

        df_progressiva_y[i] = (f(x0, y0 + deltay[i]) - f(x0, y0)) / deltay[i]
        df_regressiva_y[i] = (f(x0, y0) - f(x0, y0 - deltay[i])) / deltay[i]
        df_centrada_y[i] = (f(x0, y0 + deltay[i]) - f(x0, y0 - deltay[i])) / (2 * deltay[i])
        segunda_derivada_y[i] = (f(x0, y0 + deltay[i]) - 2 * f(x0, y0) + f(x0, y0 - deltay[i])) / (deltay[i] ** 2)

        segunda_derivada_xy[i] = (f(x0 + deltax[i], y0 + deltay[i]) - f(x0 + deltax[i], y0 - deltay[i]) - f(x0 - deltax[i], y0 + deltay[i]) + f(x0 - deltax[i], y0 - deltay[i])) / (4 * deltax[i] * deltay[i])

        erro_df_progressiva_x[i] = abs(dfx(x0, y0) - df_progressiva_x[i]) / dfx(x0, y0)
        erro_df_regressiva_x[i] = abs(dfx(x0, y0) - df_regressiva_x[i]) / dfx(x0, y0)
        erro_df_centrada_x[i] = abs(dfx(x0, y0) - df_centrada_x[i]) / dfx(x0, y0)
        erro_segunda_derivada_x[i] = abs(Dfx(x0, y0) - segunda_derivada_x[i]) / Dfx(x0, y0)

        erro_df_progressiva_y[i] = abs(dfy(x0, y0) - df_progressiva_y[i]) / dfy(x0, y0)
        erro_df_regressiva_y[i] = abs(dfy(x0, y0) - df_regressiva_y[i]) / dfy(x0, y0)
        erro_df_centrada_y[i] = abs(dfy(x0, y0) - df_centrada_y[i]) / dfy(x0, y0)
        erro_segunda_derivada_y[i] = abs(Dfy(x0, y0) - segunda_derivada_y[i]) / Dfy(x0, y0)

        erro_segunda_derivada_xy[i] = abs(Dfxy(x0, y0) - segunda_derivada_xy[i]) / Dfxy(x0, y0)

    return [
        df_progressiva_x, erro_df_progressiva_x,
        df_regressiva_x, erro_df_regressiva_x,
        df_centrada_x, erro_df_centrada_x,
        segunda_derivada_x, erro_segunda_derivada_x,

        df_progressiva_y, erro_df_progressiva_y,
        df_regressiva_y, erro_df_regressiva_y,
        df_centrada_y, erro_df_centrada_y,
        segunda_derivada_y, erro_segunda_derivada_y,

        segunda_derivada_xy, erro_segunda_derivada_xy
    ]
```

```python
def f(x):
    return 10 * np.cos(x)

def df(x):
    return - 10 * np.sin(x)

def Df(x):
    return - 10 * np.cos(x)

x0 = np.pi / 10

delta = [0.2, 0.1, 0.05, 0.025, 0.0125]
```

```python
resultado_diferenciacao = diferenciacao_numerica(x0, f, df, Df, delta)

df_progressiva = resultado_diferenciacao[0]
erro_df_progressiva = resultado_diferenciacao[1]

df_regressiva = resultado_diferenciacao[2]
erro_df_regressiva = resultado_diferenciacao[3]

df_centrada = resultado_diferenciacao[4]
erro_df_centrada = resultado_diferenciacao[5]

print(
    f"{'Δx':^10}"
    f"{'Dif. Progressiva':^20}"
    f"{'Erro Relativo':^20}"
    f"{'Dif. Regressiva':^20}"
    f"{'Erro Relativo':^20}"
    f"{'Dif. Centrada':^20}"
    f"{'Erro Relativo':^20}"
)

print("-" * 120)

for i in range(len(delta)):

    erro_progressiva_percentual = erro_df_progressiva[i] * 100
    erro_regressiva_percentual = erro_df_regressiva[i] * 100
    erro_centrada_percentual = erro_df_centrada[i] * 100

    print(
        f"{delta[i]:^10.4f}"
        f"{df_progressiva[i]:^20.4f}"
        f"{f'{erro_df_progressiva[i]:.4f} ({erro_progressiva_percentual:.2f}%)':^20}"
        f"{df_regressiva[i]:^20.4f}"
        f"{f'{erro_df_regressiva[i]:.4f} ({erro_regressiva_percentual:.2f}%)':^20}"
        f"{df_centrada[i]:^20.4f}"
        f"{f'{erro_df_centrada[i]:.4f} ({erro_centrada_percentual:.2f}%)':^20}"
    )
```

```python
def f(x, y):
    return np.exp(x * y) + np.sin(x + y)


def dfx(x, y):
    return y * np.exp(x * y) + np.cos(x + y)


def dfy(x, y):
    return x * np.exp(x * y) + np.cos(x + y)


def Dfx(x, y):
    return (y ** 2) * np.exp(x * y) - np.sin(x + y)


def Dfy(x, y):
    return (x ** 2) * np.exp(x * y) - np.sin(x + y)


def Dfxy(x, y):
    return (1 + x * y) * np.exp(x * y) - np.sin(x + y)

x0 = np.pi / 6
y0 = np.pi
```

```python
resultado_diferenciacao = diferenciacao_mista(
    x0, y0,
    f, dfx, dfy,
    Dfx, Dfy, Dfxy,
    delta, delta
)

df_progressiva_x = resultado_diferenciacao[0]
erro_df_progressiva_x = resultado_diferenciacao[1]

df_regressiva_x = resultado_diferenciacao[2]
erro_df_regressiva_x = resultado_diferenciacao[3]

df_centrada_x = resultado_diferenciacao[4]
erro_df_centrada_x = resultado_diferenciacao[5]

segunda_derivada_x = resultado_diferenciacao[6]
erro_segunda_derivada_x = resultado_diferenciacao[7]

df_progressiva_y = resultado_diferenciacao[8]
erro_df_progressiva_y = resultado_diferenciacao[9]

df_regressiva_y = resultado_diferenciacao[10]
erro_df_regressiva_y = resultado_diferenciacao[11]

df_centrada_y = resultado_diferenciacao[12]
erro_df_centrada_y = resultado_diferenciacao[13]

segunda_derivada_y = resultado_diferenciacao[14]
erro_segunda_derivada_y = resultado_diferenciacao[15]

segunda_derivada_xy = resultado_diferenciacao[16]
erro_segunda_derivada_xy = resultado_diferenciacao[17]
```
