## Métodos de Integração Numérica

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Trapézio Generalizado

```python
def f(x):
    return np.exp(x) * np.cos(x)

a = 0
b = 1.2
N = 6
n = N + 1

h = (b - a) / N
xi = np.linspace(a, b, n)
yi = f(xi)

soma = 0
for k in range(n):
    if k == 0 or k == N:
        soma = soma + yi[k]
    else:
        soma = soma + 2 * yi[k]

I_trap = (h / 2) * soma

x_fino = np.linspace(a, b, 100)
y_fino = f(x_fino)

for k in range(N):
    xp = [xi[k], xi[k], xi[k + 1], xi[k + 1]]
    yp = [0, yi[k], yi[k + 1], 0]
    plt.fill(xp, yp, color='lightblue', alpha=0.5)

for k in range(n):
    plt.plot([xi[k], xi[k]], [0, yi[k]])

plt.plot(xi, yi, 'o')
plt.plot(x_fino, y_fino, 'r-')

plt.grid(True)
plt.show()
```

### Método 1/3 Simpsom
