### 1. Aproximação da Área de uma Semicircunferência

Considere a semicircunferência de raio $r = 1$ centrada na origem, parametrizada por

$$
f(x) = \sqrt{1 - x^2}, \quad x \in [-1,1].
$$

Sabe-se que a área exata sob essa curva é igual à metade da área do círculo unitário, isto é,

$$
A = \int_{-1}^{1} \sqrt{1 - x^2}\, dx = \frac{\pi}{2}.
$$

Utilize os métodos numéricos de integração estudados para aproximar o valor da área desta semicircunferência, adotando um número de subintervalos iguais a $6$ e $12$ em cada método.

Para cada método e número de subintervalos, calcule o valor do erro absoluto obtido e apresente todos os resultados em uma tabela.

```python
import numpy as np
```

```python
def trapezio_generalizada(a, b, N, f):
    n = N + 1

    h = (b - a) / N
    xi = np.linspace(a, b, n)
    yi = f(xi)

    soma = 0
    for k in range(n):
        if k == 0 or k == N:
            soma = soma + yi[k]
        else:
            soma = soma + 2 * yi[k]

    I_trap = (h / 2) * soma

    return I_trap

def simpson_1_3(a, b, N, f):
    n = 2 * N + 1

    h = (b - a) / (2 * N)
    xi = np.linspace(a, b, n)
    yi = f(xi)

    soma = 0
    for k in range(n):
        if k == 0 or k == 2 * N:
            soma = soma + yi[k]
        elif k % 2 == 1:
            soma = soma + 4 * yi[k]
        else:
            soma = soma + 2 * yi[k]

    I_trap = (h / 3) * soma
    return I_trap

def simpson_3_8(a, b, N, f):
    n = 3 * N + 1

    h = (b - a) / (3 * N)
    xi = np.linspace(a, b, n)
    yi = f(xi)

    soma = 0
    for k in range(n):
        if k == 0 or k == 3 * N:
            soma = soma + yi[k]
        elif k % 3 == 0:
            soma = soma + 2 * yi[k]
        else:
            soma = soma + 3 * yi[k]

    I_trap = ((3 * h) / 8) * soma
    return I_trap
```

```python
def f(x):
    return np.sqrt(1 - x ** 2)
```

```python
valor_exato = np.pi / 2

metodos = [
    "Trapézio Generalizada",
    "Simpson 1/3",
    "Simpson 3/8"
]

subintervalos = [6, 12]
```

```python
aproximacoes = [
    trapezio_generalizada(-1, 1, 6, f),
    trapezio_generalizada(-1, 1, 12, f),

    simpson_1_3(-1, 1, 6, f),
    simpson_1_3(-1, 1, 12, f),

    simpson_3_8(-1, 1, 6, f),
    simpson_3_8(-1, 1, 12, f)
]
```

```python
print(f"{'Método':<25} {'Subintervalos':<18} {'Área Aproximada':<22} {'Valor Exato':<18} {'Erro Absoluto'}")
print(100 * "-")
indice = 0

for metodo in metodos:
    for n in subintervalos:

        area_aprox = aproximacoes[indice]
        erro_absoluto = abs(valor_exato - area_aprox)

        print(f"{metodo:<25} {n:<18} {area_aprox:<22.10f} {valor_exato:<18.10f} {erro_absoluto:.10f}")

        indice += 1
```
