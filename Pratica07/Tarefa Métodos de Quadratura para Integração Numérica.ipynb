### 1. Calcule

$$
\int_{1}^{3}
x^2 + 1 \, dx,
$$

com $N = 2$.

Como a integral possui intervalo definido e finito, devemos aplicar a Quadratra de Gauss-Legendre. Dessa forma, devemos transformar o intervalo $[1, 3]$ em $[-1, 1]$, fazemos isso utilizando a mudaça de variável $x = t + 2$, então $dx = dt$. Com isso temos:

$$
f(x) = x^2 + 1 \Rightarrow f(t) = (t + 2)^2 + 1
$$



```python
def f(x): 
    return (x + 2) ** 2 + 1
```

```python
integral = quadratura_de_gauss(f, 2, "Legendre")
print(f"Valor da Integral: {integral}")
```

### 2. Calcule

$$
\int_{-3}^{3}
\frac{x^2}{\sqrt{9 - x^2}}
\, dx,
$$

com $N = 2$.

Como o termo $\sqrt{9 - x^2}$ aparece, e ao utilizar a mudança de variável $x = 3t \Rightarrow dx = 3dt$, obtemos o seguinte termo: $\sqrt{9 - x^2} = \sqrt{9 - 9t^2}=3\sqrt{1 - t^2}$. Com isso, podemos utilizar a Quadratura de Gauss-Chebyshev.
$$
\int_{-3}^{3}
\frac{x^2}{\sqrt{9 - x^2}}
\, dx =

\int_{-1}^{1}
\frac{f(t)}
{\sqrt{1-t^2}}
\,dt,
$$

com $f(t) = 9 t^2$

```python
def f(x):
    return 9 * x ** 2
```

```python
integral = quadratura_de_gauss(f, 2, "Chebyshev")
print(f"Valor da Integral: {integral}")
```

### 3. Calcule

$$
\int_{2}^{+\infty}
e^{-3(x-2)} x \, dx,
$$

com $N = 3$.

A função da integral e o intervalo indicam a utilização da Qadratura de Gauss-Laguerre. Utilizando a substituição $t = 3(x - 2) \Rightarrow x = \frac{t}{3} + 2$, então $dx = \frac {1}{3}dt$.

$$
f(x) = e^{-3(x-2)} x \Rightarrow f(t) = \frac{t}{9} + \frac{2}{3}.
$$

```python
def f(x):
    return (x / 9) + (2 / 3)
```

```python
integral = quadratura_de_gauss(f, 2, "Laguerre")
print(f"Valor da Integral: {integral}")
```

### 4. Calcule

$$
\int_{-\infty}^{+\infty}
e^{-2x^2} x^2 \, dx,
$$

com $N = 2$.

Como a integral possui intervalo $[-\infty, +\infty]$, utilizamos a Quadratura de Gauss-Hermite. Utilizando a mudança de variável $t = \sqrt{2}x$, temos $x = \frac{t}{\sqrt{2}} \Rightarrow dx=\frac{1}{\sqrt{2}}dt$. Portanto:

$$
f(x) = e^{-2x^2} x^2 \Rightarrow f(t) = \frac{t^2}{2\sqrt2} 
$$

```python
def f(x):
    return (x ** 2) / (2 * np.sqrt(2))
```

```python
integral = quadratura_de_gauss(f, 2, "Hermite")
print(f"Valor da Integral: {integral}")
```
