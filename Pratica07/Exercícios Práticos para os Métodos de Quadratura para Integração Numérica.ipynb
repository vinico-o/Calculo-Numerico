$$\int_{0}^{+\infty} e^{-x} \cos(x) \, dx$$

### Exercício 2
Calcule:

$$\int_{-2}^{2} \frac{t^3 + 2t^2}{4\sqrt{4 - t^2}} \, dt$$

$N = 2$ (Responda por que a escolha de $N = 2$ é apropriada).

Como aparece o termo

$$
\sqrt{4-t^2},
$$

fazemos a mudança de variável

$$
t=2x
$$

Logo,

$$
dt=2dx
$$

e

$$
\sqrt{4-t^2}
=
\sqrt{4-4x^2}
=
2\sqrt{1-x^2}.
$$

Substituindo na integral:

$$
I
=
\int_{-1}^{1}
\frac{(2x)^3+2(2x)^2}
     {4(2\sqrt{1-x^2})}
\cdot 2\,dx
$$

$$
I
=
\int_{-1}^{1}
\frac{8x^3+8x^2}
     {4\sqrt{1-x^2}}
\,dx
$$

$$
I
=
\int_{-1}^{1}
\frac{2x^3+2x^2}
     {\sqrt{1-x^2}}
\,dx.
$$

Esta é exatamente a forma

$$
\int_{-1}^{1}
\frac{f(x)}
{\sqrt{1-x^2}}
\,dx,
$$

portanto usamos **Gauss-Chebyshev**.

A função é

$$
f(x)=2x^3+2x^2.
$$

Como Gauss-Chebyshev com $N=2$ integra exatamente polinômios até grau

$$
2N-1=3,
$$

e $f(x)$ é cúbica, a escolha de $N=2$ é apropriada.

```python
def f(x):
    return 2 * x ** 3 + 2 * x ** 2
```

```python
quadratura = quadratura_de_gauss(f, 2, "Chebyshev")
print(f"Quadratura de Gauss Chebyshev: {quadratura}")
```

### Exercício 3

Calcule

$$
I=
\int_{-\infty}^{+\infty}
\frac{e^{-x^2}}{2}x^2\,dx.
$$

$N = 2$ (Responda por que a escolha de $N = 2$ é apropriada).

Escrevendo:

$$
I=
\frac12
\int_{-\infty}^{+\infty}
e^{-x^2}x^2\,dx.
$$

A forma coincide com a quadratura de **Gauss-Hermite**

$$
\int_{-\infty}^{+\infty}
e^{-x^2}f(x)\,dx.
$$

Logo,

$$
f(x)=\frac{x^2}{2}.
$$

A função $f(x)$ é um polinômio de grau 2.

Gauss-Hermite com $N=2$ integra exatamente polinômios até grau

$$
2N-1=3,
$$

portanto $N=2$ é suficiente.

```python
def f(x):
    return x ** 2 / 2
```

```python
quadratura = quadratura_de_gauss(f, 2, "Hermite")
print(f"Quadratura de Gauss Hermite: {quadratura}")
```
